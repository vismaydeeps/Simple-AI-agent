```python
import os #using to save the api key of groq in .env
from llama_index.core.agent import ReActAgent #reasoning and acting agent. it will reason the q, understand what it has to do, what tools to use and take action based on tools and flows it has
from llama_index.llms.groq import Groq #groq llm
from llama_index.core.tools import FunctionTool #using functiontool to send the custom tools written to llm and agent
```

```python
from google.colab import userdata
userdata.get('GROQ_API_KEY')
```

Loading Groq LLM

```python
llm = Groq(
    model = "llama-3.1-8b-instant",
    temperature = 0
)
```

Creating basic arithmetic tools

```python
def add(a: float, b: float) -> float:
  """ Add two numbers and return the sum."""
  return a + b

def subtract(a: float, b: float) -> float:
  """ subtract second number from the first number and return the result."""
  return a - b

def multiply(a: float, b: float) -> float:
  """ multiply two numbers and return the product."""
  return a * b

def divide(a: float, b: float) -> float:
  """ Divide the first number by second number and return the quotient."""
  return a / b
```

```python
add_tool = FunctionTool.from_defaults(fn=add)
subtract_tool = FunctionTool.from_defaults(fn=subtract)
multiply_tool = FunctionTool.from_defaults(fn=multiply)
divide_tool = FunctionTool.from_defaults(fn=divide)
```

```python
tools_list = [add_tool, subtract_tool, multiply_tool, divide_tool]
```

Creating the agent

```python
agent = ReActAgent.from_tools(
    tools_list,
    llm = llm,
    verbose = True
)
```

**Agent in Action**

```python
response = agent.chat("what is 20 divided by 2")
```
